# Q1

```python
import numpy as np
a = np.arange(1,13).reshape(3,4)
print("数组维度(ndim):",a.ndim)
print("形状(shape):",a.shape)
print("元素总个数(size):",a.size)
print("数组内容:\n",a)
```

# Q2

```python
import numpy as np
b_float = np.array([1.5,2.6,3.7,4.8])
print("转换前数组:",b_float)
print("转换前数据类型:",b_float.dtype.name)
b_int = b_float.astype(np.int32)
print("转换后数组:",b_int)
print("转换后数据类型:",b_int.dtype.name)
```

# Q6

```python
x = np.array([1,4,9,16])
y = np.array([2,1,4,5])
sqrt_x = np.sqrt(x)
print("x的平方根:", sqrt_x)
a = np.maximum(sqrt_x, y)
print("对应位置最大值:", a)
```

# Q9

```python
arr9 = np.array([[6, 2, 7], [3, 6, 2], [4, 3, 2]])
mean_val = arr9.mean()
print("所有元素平均值:", mean_val)

arr9.sort(axis=0)
print("按列排序后数组:\n", arr9)
```
